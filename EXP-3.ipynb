```python
df = pd.read_csv('business_dataset_cleaned.csv')

print(df.isnull().sum())
print('-'*40)
print(df.duplicated().sum())
print('-'*40)
print(df.dtypes)
print('-'*40)
print(df.shape)
```

```python
df.describe()
```

```python
total_quantity = df['Quantity'].sum()
df['Revenue'] = df['Quantity'] * df['UnitPrice']
total_revenue = df['Revenue'].sum()
average_unit_price = df['UnitPrice'].mean()
total_transactions = df['InvoiceNo'].nunique()
unique_customers = df['CustomerID'].nunique()
```

```python
KPI_summary = {
    "Total Quantity sold" : total_quantity,
    "Total Revenue" : total_revenue,
    "Average Unit Price" : average_unit_price,
    "Total Unique Transactions" : total_transactions,
    "Total Unique Customers" : unique_customers
}
for kpi, value in KPI_summary.items():
    print(kpi, ":", value)

```

```python
pd.DataFrame([KPI_summary]).to_csv(
    "KPI_Summary.csv", index=False
)
```
