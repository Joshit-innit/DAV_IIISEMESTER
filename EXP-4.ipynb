```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All required libraries are imported successfully.")
```

```python
df = pd.read_csv("Superstore_Sales_Dataset.csv")
df.head()
```

```python
df.info()
```

```python
df['Order ID'] = df['Order ID'].astype(str)
df.info()
```

```python
cols = ['Region', 'Category', 'Sub-Category']
for col in cols:
    print(f"Unique values in {col} : {df[col].unique()}")
```

```python
# number of orders in each region

print("Number of orders in each region:")
region_count = df['Region'].value_counts()
plt.bar(region_count.index, region_count.values)
plt.title("Number of orders in each region")
plt.xlabel("Region")
plt.ylabel("Number of orders")
plt.show()
```

```python
# Number of orders in each category

category_count = df['Category'].value_counts()
plt.bar(category_count.index, category_count.values, )
plt.yticks(range(0, max(category_count.values) + 50, 50))
plt.title("Number of orders in each category")
plt.xlabel("Category")
plt.ylabel("Number of orders")
plt.show()
```

```python
# Number of orders in each category

subcategory_count = df['Sub-Category'].value_counts()
plt.figure(figsize=(10, 10))
plt.bar(subcategory_count.index, subcategory_count.values, )
plt.yticks(range(0, max(subcategory_count.values) + 50, 10))
plt.title("Number of orders in each sub-category")
plt.xticks(rotation=90)
plt.xlabel("Sub-Category", )
plt.ylabel("Number of orders")
plt.show()
```
