```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)

n = 1000 # number of customers
df = pd.DataFrame({
    'CustomerID' : range(1, n+1),
    'Tenure' : np.random.randint(1, 60, n),
    'MontlyCharges' : np.random.uniform(20, 100, n),
    "SupportCalls": np.random.randint(0, 10, n),
    "LoginDays": np.random.randint(0, 30, n),
    "SatisfactionScore": np.random.randint(1, 6, n),
    "Complaints": np.random.randint(0, 5, n),
    "Churn": np.random.choice(
        ["Yes", "No"],
        size=n,
        p=[0.22, 0.78]
    )

})
df.head(10)
```

```python
# customer churn rate ---> lagging KPI
churn_rate = (df['Churn'] == 'Yes').mean() * 100
print(f"Customer Churn rate: {churn_rate:.2f}%")

# Average satisfaction score ---> Leading KPI
avg_satisfaction_score = df['SatisfactionScore'].mean()
print(f"Average Satisfaction Score: {avg_satisfaction_score:.2f}")

# Average Monthly charges ---> Leading KPI
avg_monthly_charges = df['MontlyCharges'].mean()
print(f"Average Monthly charges: {avg_monthly_charges:.2f}")

# Average SupportCalls ---> Leading KPI
avg_calls = df['SupportCalls'].mean()
print(f"Average Support Calls: {avg_calls:.2f}")

# Average LoginDays ---> Leading KPI
avg_login_days = df['LoginDays'].mean()
print(f"Average Login Days: {avg_login_days:.2f}")

# total Complaints ---> Leading KPI
total_complaints = df['Complaints'].sum()
print(f"Total complaints recieved: {total_complaints:.2f}")

```

```python
kpi_dashboard = pd.DataFrame({
    "KPI" : [
        "Churn Rate",
        "Average Satisfaction Score",
        "Average Monthly Charges",
        "Average Support Calls",
        "Average Login Days",
        "Total Complaints"
    ], 
    "Values" : [
        f"{churn_rate:.2f}%",
        round(avg_satisfaction_score, 2),
        f"${avg_monthly_charges:.2f}",
        round(avg_calls, 2),
        round(avg_login_days, 2),
        total_complaints
    ]
})

print(kpi_dashboard)
```

```python
target = {
    "Churn Rate" : 15, 
    "Average Satisfaction" : 3.2,
    "Average Monthly charges" : 70,
    "Average Calls" : 3,
    "Average Login Days" : 18,
    "Total complaints" : 1200
}
print(target)

print(f"Churn Rate          : {churn_rate:.2f}% (Target < {target['Churn Rate']}%)")
print(f"Satisfaction Score  : {avg_satisfaction_score:.2f} (Target > {target['Average Satisfaction']})")
print(f"Support Calls       : {avg_calls:.2f} (Target < {target['Average Calls']})")
print(f"Login Days          : {avg_login_days:.2f} (Target > {target['Average Login Days']})")
```

```python
high_risk = df[
    (df["SatisfactionScore"] <= 2) &
    (df["SupportCalls"] >= 5) &
    (df["LoginDays"] < 10)
]

print("High Risk Customers")


print(f"\nTotal High Risk Customers: {len(high_risk)}")
```
