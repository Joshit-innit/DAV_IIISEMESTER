```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df.info()
```

```python
monthly_sales = (
    df.groupby(df["Order Date"].dt.to_period("M"))["Sales"]
      .sum()
      .reset_index()
)

monthly_sales["Order Date"] = monthly_sales["Order Date"].dt.to_timestamp()

monthly_sales
```

```python
# Step 5: Calculate the Baseline Sales
# The average monthly sales is used as a baseline for comparison.
baseline = monthly_sales["Sales"].mean()

print("Average Monthly Sales:", baseline)
```

```python
# Step 6: Create the Monthly Sales Time-Series Chart
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales["Order Date"],
    monthly_sales["Sales"],
    marker="o",
    linewidth=2
)

plt.axhline(
    y=baseline,
    linestyle="--",
    linewidth=1.5,
    label=f"Average Sales = {baseline:,.0f}"
)

plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True, alpha=0.3)
plt.legend()

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Step 7: Identify Significant Changes in Sales
# Calculate the month-to-month percentage change.
monthly_sales["Percentage Change"] = (
    monthly_sales["Sales"].pct_change() * 100
)

monthly_sales
# Display the months with the largest changes:
significant_changes = monthly_sales[
    monthly_sales["Percentage Change"].abs() >= 5
]

significant_changes

```

```python
# Step 8: Highlight Significant Changes
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales["Order Date"],
    monthly_sales["Sales"],
    marker="o",
    linewidth=2
)

plt.axhline(
    baseline,
    linestyle="--",
    linewidth=1.5,
    label="Average Sales"
)

for i, row in monthly_sales.iterrows():
    if abs(row["Percentage Change"]) >= 5:
        plt.annotate(
            f"{row['Percentage Change']:.1f}%",
            (row["Order Date"], row["Sales"]),
            xytext=(0, 10),
            textcoords="offset points",
            ha="center"
        )

plt.title("Monthly Sales Trend with Significant Changes")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True, alpha=0.3)
plt.legend()

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
